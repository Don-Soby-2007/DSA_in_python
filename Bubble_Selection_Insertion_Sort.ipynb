```python
def bubbleSort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1] , arr[j]
                swapped = True

        if swapped == False:
            break

def insertionSort(arr):
    n = len(arr)

    for i in range(1, n):
        current = arr[i]
        j = i - 1

        while j>=0 and current < arr[j]:
            arr[j+1] = arr[j]
            j -= 1

        arr[j+1] =  current

def selectionSort(arr):
    n = len(arr)

    for i in range(n - 1):

        mins = i

        for j in range(i, n):
            if arr[j] < arr[mins]:
                mins = j

        arr[i], arr[mins] = arr[mins], arr[i]
```

```python
arr = [5, 2, 9, 1, 5, 6, 10, 3]
bubbleSort(arr)
print(arr)
```
