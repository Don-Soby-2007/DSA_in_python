```python
class Node:

    def __init__(self, value, next=None, prev=None):
        self.value = value
        self.next = next
        self.prev = prev
```

```python
class DoublyLL:

    def __init__(self):
        self.head = Node(None)
    
    def display(self):
        curr = self.head.next
        arr = []

        while curr:
            arr.append(str(curr.value))
            curr = curr.next
        
        print('Head', '->', ' <-> '.join(arr), end='')
        print('\n')
    
    def append_right(self, value):
        next_node = Node(value)
        curr = self.head

        while curr.next:
            curr = curr.next
        
        curr.next, next_node.prev = next_node, curr

    def append_left(self, value):
        new_node = Node(value)
        curr = self.head
        next_node = self.head.next

        curr.next, next_node.prev, new_node.next = new_node, new_node, next_node


    def arr_dll(self, arr):
        curr = self.head

        while curr.next:
            curr = curr.next
        
        if curr == self.head:
            curr.next = Node(arr[0])
            curr = curr.next

            for i in arr[1:]:
                next_node = Node(i)
                curr.next, next_node.prev = next_node, curr
                curr = curr.next
        else:
            for i in arr:
                next_node = Node(i)
                curr.next, next_node.prev = next_node, curr
                curr = curr.next

    def pop_left(self):

        head = self.head
        curr = head.next

        head.next, curr.next.prev = curr.next, None
        del curr
    
    def pop_right(self):

        curr = self.head

        while curr.next:
            curr = curr.next

        curr.prev.next, curr.prev = None, None

        del curr


    
    
```

```python
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]

dll = DoublyLL()

dll.arr_dll(arr)
dll.display()

dll.append_left(10)
dll.append_right(20)
dll.display()

dll.pop_left()
dll.pop_right()
dll.display()
```
