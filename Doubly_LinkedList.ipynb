```python
class Node:

    def __init__(self, value, next=None, prev=None):
        self.value = value
        self.next = next
        self.prev = prev
```

```python
class DoublyLL:

    def __init__(self):
        self.head = Node(None)
    
    def display(self):
        curr = self.head.next
        arr = []

        while curr:
            arr.append(str(curr.value))
            curr = curr.next
        
        print('Head', '->', ' <-> '.join(arr), end='')
    
    def append(self, value):
        next_node = Node(value)
        curr = self.head

        while curr.next:
            curr = curr.next
        
        curr.next, next_node.prev = next_node, curr

    def arr_dll(self, arr):
        curr = self.head

        while curr.next:
            curr = curr.next
        
        if curr == self.head:
            curr.next = Node(arr[0])
            curr = curr.next

            for i in arr[1:]:
                next_node = Node(i)
                curr.next, next_node.prev = next_node, curr
                curr = curr.next
        else:
            for i in arr:
                next_node = Node(i)
                curr.next, next_node.prev = next_node, curr
                curr = curr.next
    
```

```python
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]

dll = DoublyLL()

dll.arr_dll(arr)
dll.display()
```
