```python
class Node:

    def __init__(self, value, next=None):
        self.value = value
        self.next = next
```

```python
class LinkedList:

    def __init__(self):
        self.head = Node(None)
    
    def append(self, value):
        new_node = Node(value)

        curr = self.head

        while curr.next:
            curr = curr.next
        
        curr.next = new_node
    
    def insert(self, value, pos):
        new_node = Node(value)
        curr = self.head
        i = 0

        while curr.next and i < pos-1:
            curr = curr.next
            i += 1

        curr.next, new_node.next = new_node, curr.next

    def pop_right(self):

        curr = self.head
        prev = None

        while curr.next:
            prev = curr
            curr = curr.next

        if curr != self.head:
            prev.next = None
            del curr

    def pop_left(self):

        head = self.head
        curr = head.next
        head.next = curr.next
        del curr
        
    def traversal(self):
        
        li = []

        curr = self.head

        while curr:
            li.append(str(curr.value))
            curr = curr.next
        
        return li
    
    def delete(self, value):
        if str(value) in self.traversal():
            curr = self.head
            prev = None

            while curr.next:
                if curr.value == value:
                    break
                prev = curr
                curr = curr.next
            
            if curr.next:
                prev.next = curr.next
                del curr
            else:
                prev.next = None
                del curr
    
    def display(self):
        li = self.traversal()
        print( ' -> '.join(li))
        print('---- ---- ----- -----')

    def append_first(self, value):

        head = self.head
        new_node = Node(value)

        head.next, new_node.next = new_node, head.next

    def reverse(self):

        curr = self.head.next
        prev = None

        while curr:
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node

        self.head.next = prev

    def link_set(self):

        sets = []
        curr = self.head
        prev = None

        while curr:
            if curr.value not in sets:
                sets.append(curr.value)
                prev = curr
                curr = curr.next
            else:
                prev.next = curr.next
                curr = curr.next

    def mid_node(self):

        fast = self.head.next
        slow = self.head.next

        while fast and fast.next:
            fast = fast.next.next
            slow = slow.next

        print('The middle value: ', slow.value)
```
