```python
def fact(n):

    if n == 1:
        return 1
    
    return n * fact(n-1)
```

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):

    if n <=1:
        return n
    
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
def bin_search(arr, high, low, target):

    if low > high:
        return -1
    
    mid = low + (high-low)//2
    
    if arr[mid] == target:
        return mid
    
    elif arr[mid] > target:
        return bin_search(arr, mid-1, low, target)
    else:
        return bin_search(arr, high, mid+1, target)
```

```python
arr = [12,23,45,56,67,89]

print(bin_search(arr, len(arr)-1, 0, 67))
```

```python
memo = {}
def fibonacci(n):
    # Base cases: the first two numbers are 0 and 1
    if n in memo:
        return memo[n]
    
    if n <= 1:
        return n
    
    memo[n] = fibonacci(n - 1) + fibonacci(n - 2)

    return memo[n]

# Define how many terms you want to print
terms = 10

print(f"Fibonacci series up to {terms} terms:")
for i in range(terms):
    print(fibonacci(i), end=" ")
```
