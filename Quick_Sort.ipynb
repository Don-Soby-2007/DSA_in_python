```python
# Hoare Partition

def partition1(a, low, high):

    pivot_idx = low + (high-low)//2
    pivot = a[pivot_idx]

    a[low], a[pivot_idx] = a[pivot_idx], a[low]

    start = low + 1
    end = high

    while True:
        while start <= end and a[start] <= pivot:
            start += 1
        while start <= end and a[end] > pivot:
            end -= 1

        if start < end:
            a[start], a[end] = a[end], a[start]
        else:
            break

    a[low], a[end] = a[end], a[low]

    return end

# Lomuto Partition

def partition2(a, low, high):

    pivot = a[high]
    i = low - 1

    for j in range(low, high):
        if a[j] < pivot:
            i +=1
            a[i], a[j] = a[j], a[i]

    a[i+1], a[high] = a[high] , a[i+1]

    return i+1
```

```python
def quicksort(a, low, high):
    if low < high:
        pos = partition2(a, low, high)

        quicksort(a, low, pos - 1)
        quicksort(a, pos + 1, high)
```

```python
arr = [5, 10, 2, 3, 4, 5, 3, 8, 9, 3, 5, 6]
quicksort(arr, 0, len(arr)-1)
print(arr)
```
