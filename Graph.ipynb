```python
from collections import deque

class Graph:

    def __init__(self):
        self.graph = {}

    def add_node(self, node):

        if node not in self.graph:
            self.graph[node] = []

    def add_edge(self, u, v):

        self.add_node(u)
        self.add_node(v)

        self.graph[u].append(v)
        self.graph[v].append(u)

    def display(self):

        for node, neig in self.graph.items():

            print(f"{node} -> {neig}")

    def has_cycle(self):

        visited = set()

        for node in self.graph:

            if node not in visited:

                if self._is_cycle(node, visited, None):
                    return True
                
        return False
    
    def _is_cycle(self, node, visited, parent):

        visited.add(node)

        for neighbor in self.graph[node]:

            if neighbor not in visited:

                if self._is_cycle(neighbor, visited, node):
                    return True
                
            elif neighbor != parent:
                return True
                
        return False
    
    def degree_of_seperation(self, start, target):

        visited = set()

        q = deque()

        q.append((start, 0))

        visited.add(start)

        while q:

            node, distance = q.popleft()

            if node == target:
                return distance
            
            for neig in self.graph[node]:

                if neig not in visited:

                    visited.add(neig)

                    q.append((neig, distance + 1))


        return -1
    
    
```

```python
g = Graph()

g.add_edge("A", "B")
g.add_edge("B", "C")
g.add_edge("C", "D")
g.add_edge("D", "A")

g.display()

print(g.has_cycle())
print(g.degree_of_seperation('A', 'D'))
```
