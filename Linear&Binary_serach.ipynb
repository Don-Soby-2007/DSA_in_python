```python
def lin_serch(value, arr):

    for i in range(len(arr)):
        if arr[i] == value:
            return i
        
    return -1


def bin_serach(value, arr):

    low = 0
    high = len(arr) - 1
    
    while high > low:

        if arr[low] == value:
            return low
        
        if arr[high] == value:
            return high

        mid = low + (high-low)//2

        if arr[mid] == value:
            return mid
        
        if value > arr[mid]:
            low = mid + 1
        else:
            high = mid - 1
    
    return -1
```

```python
arr = [10, 20, 30, 40, 50, 60]
ar = [54, 21, 32, 78, 65, 53, 75]

pos1 = lin_serch(32, ar)
pos2 = bin_serach(60, arr)

print(pos1, pos2)
```
