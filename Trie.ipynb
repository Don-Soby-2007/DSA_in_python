```python
class TrieNode:

    def __init__(self):
        self.children = [None] * 26
        self.isEndOfWord = False
```

```python
class Trie:

    def __init__(self):
        self.root = TrieNode()

    def insert(self, value):

        curr = self.root

        for i in value:

            index = ord(i) - ord('a')

            if curr.children[index] is None:

                new_node = TrieNode()

                curr.children[index] = new_node

            curr = curr.children[index]

        curr.isEndOfWord = True

    def search(self, key):

        curr = self.root

        for i in key:

            index = ord(i) - ord('a')

            if curr.children[index]:
                curr = curr.children[index]
            else:
                return False
                

        return curr.isEndOfWord
        

    def isPrefix(self, key):

        curr = self.root

        for i in key:

            index = ord(i) - ord('a')

            if curr.children[index]:
                curr = curr.children[index]
            else:
                return False
            
        return True
    
```

```python
trie = Trie()
arr = ["and", "ant", "do", "dad"]
for s in arr:
    trie.insert(s)
searchKeys = ["do", "gee", "bat"]
for s in searchKeys:
    if trie.search(s):
        print("true", end= " ")
    else:
        print("false", end=" ")

print()
prefixKeys = ["ge", "ba", "do", "de", "nt"]
for s in prefixKeys:
    if trie.isPrefix(s):
        print("true", end = " ")
    else:
        print("false", end = " ")
```
