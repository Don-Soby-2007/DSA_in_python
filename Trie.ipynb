```python
class TrieNode:

    def __init__(self):
        self.children = [None] * 26
        self.isEndOfWord = False
```

```python
class Trie:

    def __init__(self):
        self.root = TrieNode()

    def insert(self, value):

        curr = self.root

        for i in value:

            index = ord(i) - ord('a')

            if curr.children[index] is None:

                new_node = TrieNode()

                curr.children[index] = new_node

            curr = curr.children[index]

        curr.isEndOfWord = True

    def search(self, key):

        curr = self.root

        for i in key:

            index = ord(i) - ord('a')

            if curr.children[index]:
                curr = curr.children[index]
            else:
                return False
                

        return curr.isEndOfWord
        

    def isPrefix(self, key):

        curr = self.root

        for i in key:

            index = ord(i) - ord('a')

            if curr.children[index]:
                curr = curr.children[index]
            else:
                return False
            
        return True
    
    def delete(self, word, curr=None, depth=0):

        if curr is None:
            curr = self.root

        if depth == len(word):

            if not curr.isEndOfWord:
                return False
            
            curr.isEndOfWord = False
            
            return self._isEmpty(curr)
        
        index = ord(word[depth]) - ord('a')

        child = curr.children[index]

        if child is None:
            return False
        
        shouldDeleteChild = self.delete(word, child, depth+1)

        if shouldDeleteChild:

            curr.children[index] = None

            return (not curr.isEndOfWord and self._isEmpty(curr))
        
        return False
    
    def _isEmpty(self, node):
        for i in node.children:

            if i is not None:
                return False
        
        return True
    
    def suggestion(self, prefix):

        curr = self.root

        for ch in prefix:

            index = ord(ch) - ord('a')

            if curr.children[index] is None:
                return []
            
            curr = curr.children[index]

        result = []

        self._dfs(curr, prefix, result)

        return result

    def _dfs(self, node, word, result):

        if node.isEndOfWord:
            result.append(word)

        for i in range(26):
            
            child = node.children[i]
            
            if child is not None:
                ch = chr(i + ord('a'))

                self._dfs(child, word + ch, result)

    
```

```python
trie = Trie()
arr = ["and", "ant", "do", "dad"]
for s in arr:
    trie.insert(s)
searchKeys = ["do", "gee", "bat"]
for s in searchKeys:
    if trie.search(s):
        print("true", end= " ")
    else:
        print("false", end=" ")

print()
prefixKeys = ["ge", "ba", "do", "de", "nt"]
for s in prefixKeys:
    if trie.isPrefix(s):
        print("true", end = " ")
    else:
        print("false", end = " ")
```

```python
trie1 = Trie()

trie1.insert("cat")
trie1.insert("car")
trie1.insert("care")
trie1.insert("camera")

print(trie1.search("care"))   # True

trie1.delete("care")

print(trie1.search("care"))   # False
print(trie1.search("ca"))    # True
print(trie1.search("cat"))
```
