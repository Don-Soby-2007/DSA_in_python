```python
def merge(arr, low, mid, high):
    
    temp = [None] * (high - low + 1)
    i = low
    j = mid + 1
    k = 0

    while i <= mid and j <= high:
        if arr[i] <= arr[j]:
            temp[k] = arr[i]
            i += 1
            k +=1
        else:
            temp[k] = arr[j]
            j += 1
            k += 1
    
    
    while i <= mid:
        temp[k] = arr[i]
        k += 1
        i += 1
    
    while j <= high:
        temp[k] = arr[j]
        k += 1
        j += 1

    if len(temp)-1 == k:
        print('K is finished')

    k = 0
    
    for i in range(low, high+1):
        arr[i] = temp[k]
        k += 1
        
```

```python
def mergeSort(arr, low, high):

    if low < high:
        mid = low + (high - low)//2

        mergeSort(arr, low, mid)
        mergeSort(arr, mid + 1, high)
        merge(arr, low, mid, high)
```

```python
arr = [1, 5, 3, 7, 8, 6, 2]
mergeSort(arr, 0, len(arr)-1)
print(arr)
```
