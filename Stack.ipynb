```python
class Node:

    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class Stack:

    def __init__(self):
        self.top = None
        self._size = 0

    def isEmpty(self):
        return self.top is None
    
    def push(self, date):

        new_node = Node(date)

        new_node.next = self.top
        self.top = new_node
        self._size += 1

    def pop(self):

        if self.isEmpty():
            raise IndexError("Pop from a empty stack")
        
        poped = self.top
        self.top = self.top.next
        self._size -= 1

        return poped.value
    
    def peek(self):
        if self.isEmpty():
            return None
        
        return self.top.value
    
    def size(self):
        return self._size
```
