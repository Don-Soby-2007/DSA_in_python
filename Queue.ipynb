```python
class CircularQueue:
    def __init__(self, k):
        self.k = k  # Fixed capacity
        self.queue = [None] * k
        self.head = -1
        self.tail = -1

    def enqueue(self, value):
        if self.is_full():
            print("Queue is Full!")
            return False
        
        if self.is_empty():
            self.head = 0
        
        # Wrap the tail around using modulo
        self.tail = (self.tail + 1) % self.k
        self.queue[self.tail] = value
        return True

    def dequeue(self):
        if self.is_empty():
            print("Queue is Empty!")
            return False
        
        result = self.queue[self.head]
        
        # If there was only one element, reset pointers
        if self.head == self.tail:
            self.head = -1
            self.tail = -1
        else:
            # Wrap the head around using modulo
            self.head = (self.head + 1) % self.k
        return result

    def is_empty(self):
        return self.head == -1

    def is_full(self):
        return (self.tail + 1) % self.k == self.head
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedQueue:
    def __init__(self):
        self.tail = None
        self._size = 0

    def is_empty(self):
        return self.tail is None

    def enqueue(self, data):
        new_node = Node(data)
        if self.is_empty():
            new_node.next = new_node  # Point to itself
            self.tail = new_node
        else:
            # New node points to the current head
            new_node.next = self.tail.next
            # Current tail points to the new node
            self.tail.next = new_node
            # Move the tail pointer to the new node
            self.tail = new_node
        self._size += 1

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Queue is empty")
        
        head = self.tail.next
        if self.tail == head:  # Only one element
            self.tail = None
        else:
            # Tail points to the node after the current head
            self.tail.next = head.next
        
        self._size -= 1
        return head.data

    def peek(self):
        if self.is_empty():
            return None
        return self.tail.next.data # The head is tail.next
```
