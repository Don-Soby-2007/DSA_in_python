```python
print(arr)

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]

    left += 1
    right -=1

print(arr)
    
```

```python
# flatten List

def flatten_array(arr):

    for i in arr:

        if isinstance(i, list):
            yield from flatten_array(i)
        else:
            yield i


arr = [[1, 2, 3, 4], [5, 6, 7, 8], [9, 0, 1, [2, 3, 4,]]]
print(arr)

arr = list(flatten_array(arr))

print(arr)
```

```python
# Array reversal

ll = [1, 2, 3, 4, 5, 6]
print(ll)

if len(ll)%2 == 0:
    for i in range(0 ,(len(ll)-1)//2):
        ll[i], ll[(len(ll)-1)-i] = ll[(len(ll)-1)-i], ll[i]

else:
    for i in range(0, (len(ll)-1)//2 - 1):
        ll[i], ll[(len(ll)-1)-i] = ll[(len(ll)-1)-i], ll[i]

print(ll)
```
