```python
arr = [1, 2, 3, 4, 5, 6, 7, 8]  

print(arr)
```

```python
arr.insert(2,4)
print(arr)
```

```python
arr.pop()
print(arr)
```

```python
arr.pop(3)
print(arr)
```

```python
max = 0

for i in arr:
    if i > max:
        max = i
    
print(max)
```

```python
min = float('inf')

for i in arr:
    if i < min:
        min = i

print(min)
```

```python
arr2 = [0] * len(arr)

for i in range(0, len(arr)):
    arr2[len(arr)-1-i] = arr[i]

print(arr2)
```

```python
print(arr)

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]

    left += 1
    right -=1

print(arr)
    
```
