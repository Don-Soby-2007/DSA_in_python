```python
class MinHeap:

    def __init__(self,):
        self.heap = []
    
    def _parent(self, i):
        return (i - 1) // 2
    
    def _left(self, i):
        return (2 * i) + 1
    
    def _right(self, i):
        return (2 * i) + 2
    
    def insert(self, value):
        self.heap.append(value)
        """ value inserting at the end of the array so we 
        want to shift it to the correct position 
        So we use buble up """
        self._bubble_up(len(self.heap) - 1)

    def _bubble_up(self, i):
        """
        It is a function used for moving upward the value to the correct position.
        it find the parent id of that index and check it is grater than the value we inserted.
        if it is it swap the element in the array
        """
        while i > 0 and self.heap[self._parent(i)] > self.heap[i]:
            parent_id = self._parent(i)
            self.heap[i], self.heap[parent_id] = self.heap[parent_id], self.heap[i]
            i = parent_id

    def extract_min(self):

        if len(self.heap) == 0:
            return None
        elif len(self.heap) == 1:
            return self.heap.pop()
        
        """
        for getting the min number you can find it in the 0th position
        when we want to delete an element that has 3 or more element we want to add
        smallest value to the up and re-order the heap

        """
        
        root = self.heap[0] 
        """ 
        root is var that holds the min value
        then we will change the first position with the last element
        So, we want to delete the element at the same time 
        that's why we use pop
        """
        self.heap[0] = self.heap.pop()

        """ 
        bubble down is the function used to find the min value in the 
        left and right hand and swap with parent
        """
        self._bubble_down(0)

        return root
    
    def _bubble_down(self, i):

        """ 
        bubble down is the function used to find the min value in the 
        left and right hand and swap with parent

        first we consider 'i' as a smallest element
        then check it with the heap left and right side

        """

        smallest = i
        left = self._left(i)
        right = self._right(i)

        if left < len(self.heap) and self.heap[left] < self.heap[smallest]:
            smallest = left
        if right < len(self.heap) and self.heap[right] < self.heap[smallest]:
            smallest = right

        """ 
        If the smallest changed this block will work and swap the child 
        with parent
        the call anothe bubble_down to do the same thing continuously 
        by achiving the element at the correct position
        
        """

        if smallest != i:
            self.heap[i], self.heap[smallest] = self.heap[smallest], self.heap[i]
            self._bubble_down(smallest)

    def display(self):
        print(self.heap)

    
    def heapify(self, arr):
        self.heap = arr
        start = (len(arr) // 2) - 1

        for i in range(start, -1, -1):
            self._bubble_down(i)

        


```

```python
arr = [1, 3, 2, 4, 4, 6, 7, 9, 12]

heap = MinHeap()

for i in arr:
    heap.insert(i)

heap.display()
print(f'Minimum number in heap : {heap.extract_min()}')
print(f'Second Minimum number in heap : {heap.extract_min()}')
heap.display()
```
